```python
import torch
import torch.nn as nn
```

```python
#input
torch.manual_seed(42)
inputs = torch.tensor([1.0,2.0,3.0])
```

```python
neuron = nn.Linear(in_features=3, out_features=1)

#output
output = neuron(inputs) # creates an object of the nn.Linear class.

output
```
